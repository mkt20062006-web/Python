```python
print("================================================================")
print("          SMARTCAMPUS UTILITY & ACCESS PASS GENERATOR          ")
print("================================================================")
import sys
user=int(input("Select User Category (1: Student, 2: Faculty/Staff):"))
base_fee=0
merit_discount=0
parking_fee=0
sur=0
if user==1:
    a=input("Enter Sub-Category (UG / PG):").upper()
    x="UG"
    y="PG"
    if a==x:
        base_fee=500
        
    elif a==y:
        base_fee=350
        
    else:
        sys.exit("Error: wrong sub-category")
elif user==2:
    a=input("Enter Sub-Category (resident faculty / guest faculty):").lower()
    x="resident faculty"
    y="guest faculty"
    if a==x:
        base_fee=800
        
    elif a==y:
        base_fee=1200
        
    else:
        sys.exit("Error: wrong sub-category")
       
else:
    sys.exit("Error: wrong category")
    
if user==1:
    cgpa=float(input("Enter Student CGPA (0.0 - 10.0): "))
    if cgpa>=8.5 and cgpa<=10:
        merit_discount=base_fee*0.20
        dis="20%"
    elif cgpa>7.49 and cgpa<8.5:
        merit_discount=base_fee*0.10
        dis="10%"
    elif cgpa<7.5 and cgpa>=0:
        merit_discount=0
        dis="0%"
    else:
        sys.exit("Error: CGPA must be (0.0 - 10.0)")
        
if user==2:
    working_year=float(input("enter your working years"))
    if working_year>10:
        merit_discount=base_fee*0.15
        dis="15%"
    elif working_year<0:
        sys.exit("error: working year cannot be negative")
    else:
        merit_discount=0
        dis="0%"
        
vehicle=int(input("Select Parking Permit (0: None, 2: Two-Wheeler, 4: Four-Wheeler):"))
if user==1:
    if vehicle==2:
        sur=0
        parking_fee=200
        v="Two -wheeler"
    elif vehicle==4:
        sur=150
        parking_fee=600
        v="Four-wheeler"
    elif vehicle==0:
        parking_fee=0
        sur=0
        v="No - vehicle"
    else:
        sys.exit("Error: invalid vehicle type")
        
elif user==2:
    if vehicle==2:
        sur=0
        parking_fee=200
        v="Two -wheeler"
    elif vehicle==4:
        sur=0
        parking_fee=600
        v="Four-wheeler"
        
    elif vehicle==0:
        parking_fee=0
        sur=0
        v="No - vehicle"
    else:
        sys.exit("Error: invalid vehicle type")
    
units=int(input("Enter Monthly Electricity Consumption (in kWh):"))
if units>=0 and units<=100:
    rate=units*3+50
    ch=50
    
elif units>100 and units<=300:
    rate=units*5+100
    ch=100
    
elif units>300 and units<=500:
    rate=units*7.50+150
    ch=150
    
elif units>500:
    rate=units*10+250
    ch=250
else:
    sys.exit("Error: unit must be positive")
    
print("----------------------------------------------------------------")
print("                     CALCULATED INVOICE BREAKDOWN                ")
print("----------------------------------------------------------------")
print(f"Base Access Pass Fee       : ₹{base_fee:.2f}")
print(f"Merit Discount ({dis})       :-₹{merit_discount:.2f}")
print(f"Parking Fee ({v}) : ₹{parking_fee:.2f}")
print(f"Student Peak Surcharge     : ₹{sur:.2f}")
print(f"Net Pass & Parking Total   : ₹{parking_fee+sur-merit_discount+base_fee:.2f}")
print("----------------------------------------------------------------")
print(f"Electricity Bill ({units} kwh)  : ₹{rate:.2f} (Slab calculated + fixed charge ₹{ch})")
print("----------------------------------------------------------------")
print(f"TOTAL MONTHLY PAYABLE      : ₹{base_fee-merit_discount+parking_fee+sur+rate:.2f}")
print("================================================================")
```
